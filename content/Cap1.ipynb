## 1. Busqueda de conetores clave
### Datos: **lista** de *usuarios* de la empresa

```python
# variable = Lista => diccionario
# variable = [{}]
# variable = [{llave: valor, llave: valor, llave: valor}, {llave: valor,llave: valor,llave: valor}, ... , {llave: valor, llave: valor, llave: valor}]
# usuarios = [{"id": "valor_id", "nombre": "valor_nombre"} ]

usuarios = [
    {"id": 0, "nombre": "Hero"},
    {"id": 1, "nombre": "Dunn"},
    {"id": 2, "nombre": "Sue"},
    {"id": 3, "nombre": "Chi"},
    {"id": 4, "nombre": "Thor"},
    {"id": 5, "nombre": "Clive"},
    {"id": 6, "nombre": "Hicks"},
    {"id": 7, "nombre": "Devin"},
    {"id": 8, "nombre": "Kate"},
    {"id": 9, "nombre": "Klein"}
]
```

### Datos: **lista** por pares de id's que representa las *amistades_pares*

```python
# variable = Lista => tupla
# variable = [()]
# variable = [(valor_1, valor_2), (valor_1, valor_2),(valor_1, valor_2), ... , (valor_1, valor_2)]
# amistades_pares = [(usuario_id, usuario_id)]

amistades_pares = [(0, 1), (0, 2), (1, 2), (1, 3), (2, 3), (3, 4),
                   (4, 5), (5, 6), (5, 7), (6, 8), (7, 8), (8,9)]
```

![red de amigos de DataScencecester](assets/images/ilustracionRedAmigos.png)

### Una lista de parejas no es la fomra mas facil de trabajar

[(0, 1), (0, 2), (1, 2), (1, 3), (2, 3), (3, 4), (4, 5), (5, 6), (5, 7), (6, 8), (7, 8), (8, 9)]

```python
usuario = 1
print("amistasdes del usuario "+ str(usuario) + ":" )

for i, j in amistades_pares:        #obtenemos los id de los usuarios ue estan en la amistad par
    if usuario in (i, j):           #haciendo comparacion => el usuario esta en esa amistad?
        if i != usuario:            #si el primer id no soy yo mismo
            print(i)                    #entonces es mi amigo
        elif j != usuario:          #si el sgundo id de la tupla no soy yo mismo
            print(j)                    #entonces es mi amigo
```

### Mejor usemos un *dict*  
#### Buscar las cosas en un *dict* es muy rapido:
---
**dict:** *{llave : valor}*  
*{ usurio_id : [amigo_id, ..., amigo_id] }*

```python
# Inicializamos el diccionario con una lista vacia para cada ID  de usuario:
# amistades = { usuario_id : [], ..., usuario_id : [] }

amistades = { usuario["id"]: [] for usuario in usuarios}
```

`amistades = { usuario["id"]: [] for usuario in usuarios}` equivale a:
----
~~~
amistades = {}
for usuario in usuarios:
    amistades[usuario['id']] = []
~~~

```python
# Ahora recorremos las parejas de amistades para llenar cada usuario con su amigo:

for i, j  in amistades_pares:
    amistades[i].append(j)          #agregamos j como amigo del usuario i
    amistades[j].append(i)          #agregamos i como amigo del usuario j
```

```python
usuario = 4
print("amistasdes del usuario "+ str(usuario) + ":" )

for i, j in amistades_pares:        #obtenemos los id de los usuarios ue estan en la amistad par
    if usuario in (i, j):           #haciendo comparacion => el usuario esta en esa amistad?
        if i != usuario:            #si el primer id no soy yo mismo
            print(i)                    #entonces es mi amigo
        elif j != usuario:          #si el sgundo id de la tupla no soy yo mismo
            print(j)                    #entonces es mi amigo
```

#### ¿Cuál es el número promedio de conexiones?

```python
#1ro - Hallamos el número total de conexiones
def numero_de_amigos(usuario):
    """¿Cúantos amigos tiene _usuario_?"""
    usuario_id = usuario["id"]              # usuario = { "id": 0, "nombre": "Hero" }
    amigos_id = amistades[usuario_id]       # amigos_id = amistades[0] => amigos_id = [1, 2]
    return len(amigos_id)                   # 2
```

```python
for num in usuarios:
    print(numero_de_amigos(num))
```

#### `total_conexiones` es igual a la **summa** del número de amigos de cada usuario

```python
total_conexiones = sum(numero_de_amigos(usuario) for usuario in usuarios)           # sum(2+3+3+3+2+3+2+2+3+1)
print(total_conexiones)
```

#### `promedio_conexiones` es igual a la **suma** del número de amigos de cada usuario **dividido** en el total de usuarios

```python
numero_usuarios = len(usuarios)                                 # 10
promedio_conexiones = total_conexiones / numero_usuarios        # 24/10
print(numero_usuarios)
print(promedio_conexiones)
```

#### ¿Quiénes son las personas más conectadas?

```python
# Creamos una lista de pares donde tengamos el usuario_id, numero_de_amigos
# num_amigos_por_id = [(usuario_id, numero_de_amigos), ..., (usuario_id, numero_de_amigos)]

num_de_amigos_por_id = [ (usuario["id"], numero_de_amigos(usuario)) for usuario in usuarios ]
print('nuemro de amigos por id: ' + str(num_de_amigos_por_id))
```

#### Ordenamos la lista de `num_amigos_por_id`

```python
num_de_amigos_por_id.sort(key= lambda ids_de_amigos : ids_de_amigos[1], reverse=True)           # ids_de_amigos = (0, 2)
print('número de amigos ordenados por cantidad de amigos: ' + '\n' + str(num_de_amigos_por_id))
```
